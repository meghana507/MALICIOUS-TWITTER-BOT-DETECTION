```python
%%writefile app.py
import streamlit as st
import pandas as pd
import joblib

# Load your saved model
with open("lr.joblib", "rb") as f:
    model = joblib.load(f)

st.title("Twitter Bot Detection")

# Upload CSV file
uploaded_file = st.file_uploader("Upload CSV with features")
if uploaded_file:
    data = pd.read_csv(uploaded_file)
    st.write("Data preview:")
    st.write(data.head())

    # Make predictions
    predictions = model.predict(data)
    st.write("Predictions:")
    st.write(predictions)
```
